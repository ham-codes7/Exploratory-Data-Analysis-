## Handling Missing Data

This section demonstrates various ways to handle missing values (NaN) in pandas DataFrames.

```python
# Import necessary libraries
import numpy as np
import pandas as pd
```

### Creating an initial DataFrame

Let's assume we have a DataFrame as shown.

```python
data = np.arange(15, 30).reshape(5, 3)
dfx = pd.DataFrame(data, index=['apple', 'banana', 'kiwi',
'grapes', 'mango'], columns=['store1', 'store2', 'store3'])
display(dfx)
```

### Adding missing values to the DataFrame

We'll introduce `NaN` values to demonstrate missing data handling.

```python
dfx['store4'] = np.nan
dfx.loc['watermelon'] = np.arange(15, 19)
dfx.loc['oranges'] = np.nan
dfx['store5'] = np.nan
dfx['store4']['apple'] = 20.
display(dfx)
```

### Checking for NaN values in pandas objects

We can use `isnull()`, `notnull()`, `sum()`, and `count()` methods to inspect missing values.

```python
# 1. Check the following example:
display(dfx.isnull())
```

```python
# 2. Check it out in action:
display(dfx.notnull())
```

```python
# 3. We can use the sum() method to count the number of NaN values in each store.
display(dfx.isnull().sum())
```

```python
# 4. We can go one level deeper to find the total number of missing values
display(dfx.isnull().sum().sum())
```

```python
# 5. So, instead of counting the number of missing values, we can count the number of reported values:
display(dfx.count())
```

### Dropping missing values

One way to handle missing values is to remove them from our dataset using `dropna()`.

```python
# Display non-null values in 'store4'
display(dfx.store4[dfx.store4.notnull()])

# Drop NaN values from 'store4'
display(dfx.store4.dropna())
```

```python
# If dropna() is applied to the entire dataframe, then it will drop all rows with at least one NaN value.
display(dfx.dropna())
```

#### Dropping by rows

We can drop rows that have entirely NaN values using `how='all'`.

```python
display(dfx.dropna(how='all'))
```

#### Dropping by columns

We can also pass `axis=1` to check for NaNs by columns, and `thresh` to specify a minimum number of non-NaN values required.

```python
# Drop columns where all values are NaN
display(dfx.dropna(how='all', axis=1))
```

```python
# Drop columns that have less than 5 non-NaN values
display(dfx.dropna(thresh=5, axis=1))
```

### Mathematical operations with NaN

NaN values can affect mathematical operations like mean, sum, and cumulative sum.

```python
ar1 = np.array([100, 200, np.nan, 300])
ser1 = pd.Series(ar1)
print(f"Numpy array mean: {ar1.mean()}, Pandas Series mean: {ser1.mean()}")
```

```python
# Let's compute the total quantity of fruits sold by store4:
ser2 = dfx.store4
print(f"Sum of store4: {ser2.sum()}")
```

```python
# Average
print(f"Average of store4: {ser2.mean()}")
```

```python
# Cumulative Summing
display(ser2.cumsum())
```

### Filling missing values

We can use the `fillna()` method to replace NaN values with specified values or using interpolation methods like forward-fill (`ffill`) or backward-fill (`bfill`).

```python
# Fill NaN values with 0
filledDf = dfx.fillna(0)
display(filledDf)
```

```python
# Check the difference in mean between NaN and 0-filled DataFrames
print("Original DataFrame mean (excluding NaN):")
display(dfx.mean())
print("Filled DataFrame mean (NaN replaced with 0):")
display(filledDf.mean())
```

#### Backward and forward filling

```python
# Forward-filling (method = 'ffill')
display(dfx.store4.fillna(method='ffill'))
```

```python
# Backward-filling (method = 'bfill')
display(dfx.store4.fillna(method='bfill'))
```

## Advanced Missing Value Imputation with scikit-learn

This section demonstrates various imputation techniques for handling missing values using `sklearn`.

```python

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer
# numpy and pandas are already imported in previous cells
```

### Create a sample DataFrame with missing values

```python
# Create sample DataFrame with missing values (NaN)
df = pd.DataFrame(
    {
        "Age": [25, np.nan, 30, 45, 22],
        "Salary": [50000, 60000, np.nan, 80000, 45000],
        "Category": ["A", "B", "A", np.nan, "B"],
    }
)
display(df)
```

### 1. Mean Imputation (for Numerical Features)

This method replaces missing numerical values with the mean of the column.

```python
# 1. Mean Imputation (Numerical)
mean_imputer = SimpleImputer(strategy="mean")
df["Age_Mean"] = mean_imputer.fit_transform(df[["Age"]])
display(df)
```

### 2. Median Imputation (for Numerical Features)

Median imputation is robust to outliers and replaces missing numerical values with the median of the column.

```python
# 2. Median Imputation (Numerical - robust to outliers)
median_imputer = SimpleImputer(strategy="median")
df["Salary_Median"] = median_imputer.fit_transform(df[["Salary"]])
display(df)
```

### 3. Mode Imputation (for Categorical Features)

Mode imputation replaces missing categorical values with the most frequent value in the column.

```python
# 3. Mode Imputation (Categorical - most frequent)
mode_imputer = SimpleImputer(strategy="most_frequent")
# The fit_transform method returns a 2D array, which causes an error when assigning to a single column.
# We need to flatten it to a 1D array.
df["Category_Mode"] = mode_imputer.fit_transform(df[["Category"]]).ravel()
display(df)
```

### 4. MICE Imputation (Multivariate Imputation by Chained Equations)

MICE is an advanced method that models each feature with missing values as a function of other features, and then imputes the missing values sequentially.

```python
# 4. MICE Imputation (Multivariate Imputation by Chained Equations)
# Select numerical columns for MICE
num_cols = ["Age", "Salary"]
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice = df[num_cols].copy()
df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)
df["Age_MICE"] = df_mice["Age"]
df["Salary_MICE"] = df_mice["Salary"]
display(df)
```
