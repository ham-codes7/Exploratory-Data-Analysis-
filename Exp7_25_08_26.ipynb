## Exploratory Data Analysis on Iris Dataset

```python
# 1. Data Loading & Inspection
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Iris dataset
iris = sns.load_dataset('iris')

print("--- First 5 rows of the dataset ---")
display(iris.head())

print("\n--- Shape of the dataset ---")
print(iris.shape)

print("\n--- Data types ---")
print(iris.info())

print("\n--- Missing values check ---")
print(iris.isnull().sum())
```

## 2. Visualizations

```python
# Histogram / Distribution Plot: Distribution of 'sepal_length'
plt.figure(figsize=(10, 6))
sns.histplot(iris['sepal_length'], kde=True, bins=20, color='skyblue')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Barplot: Average 'petal_length' for each species
plt.figure(figsize=(10, 6))
sns.barplot(x='species', y='petal_length', data=iris, palette='viridis')
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length (cm)')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Boxplot: Distribution of 'sepal_width' across species
plt.figure(figsize=(10, 6))
sns.boxplot(x='species', y='sepal_width', data=iris, palette='plasma')
plt.title('Distribution of Sepal Width by Species')
plt.xlabel('Species')
plt.ylabel('Sepal Width (cm)')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Violin Plot: Distribution and density of 'petal_width' across species
plt.figure(figsize=(10, 6))
sns.violinplot(x='species', y='petal_width', data=iris, palette='magma')
plt.title('Distribution and Density of Petal Width by Species')
plt.xlabel('Species')
plt.ylabel('Petal Width (cm)')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Jointplot: 'sepal_length' versus 'sepal_width'
sns.jointplot(x='sepal_length', y='sepal_width', data=iris, kind='scatter', color='darkgreen')
plt.suptitle('Jointplot of Sepal Length vs Sepal Width', y=1.02) # Adjust suptitle position
plt.show()
```

```python
# Regplot: Regression line comparing 'petal_length' against 'petal_width'
plt.figure(figsize=(10, 6))
sns.regplot(x='petal_length', y='petal_width', data=iris, color='purple', scatter_kws={'alpha':0.6})
plt.title('Regression Plot of Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Pairplot: Pairwise relationships of the entire dataset, color-coded by 'species'
# This plot might take a moment to generate
sns.pairplot(iris, hue='species', palette='husl')
plt.suptitle('Pair Plot of Iris Dataset by Species', y=1.02) # Adjust suptitle position
plt.show()
```

```python
# Swarmplot: Categorical scatter plot of 'petal_length' distributed by 'species'
plt.figure(figsize=(10, 6))
sns.swarmplot(x='species', y='petal_length', data=iris, palette='coolwarm')
plt.title('Petal Length Distribution by Species (Swarm Plot)')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
